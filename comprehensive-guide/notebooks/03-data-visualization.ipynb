## 📊 Statistical Visualizations

```python
# Create sample dataset
np.random.seed(42)
n = 1000

data = {
    'feature_1': np.random.normal(50, 15, n),
    'feature_2': np.random.exponential(2, n),
    'feature_3': np.random.gamma(2, 2, n),
    'category': np.random.choice(['A', 'B', 'C'], n, p=[0.5, 0.3, 0.2]),
    'target': np.random.binomial(1, 0.3, n)
}

# Add some correlation
data['feature_4'] = data['feature_1'] * 0.8 + np.random.normal(0, 5, n)
data['target'] = (data['feature_1'] > 50).astype(int) + np.random.binomial(1, 0.2, n)
data['target'] = np.clip(data['target'], 0, 1)

df = pd.DataFrame(data)
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Comprehensive EDA visualization function
def create_eda_dashboard(df, target_col=None):
    """
    Create comprehensive EDA dashboard
    """
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if target_col and target_col in numeric_cols:
        numeric_cols.remove(target_col)
    
    categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
    
    fig, axes = plt.subplots(3, 3, figsize=(20, 15))
    fig.suptitle('Comprehensive EDA Dashboard', fontsize=16, fontweight='bold')
    
    # 1. Distribution plots
    for i, col in enumerate(numeric_cols[:3]):
        axes[0, i].hist(df[col], bins=30, alpha=0.7, edgecolor='black')
        axes[0, i].set_title(f'Distribution of {col}')
        axes[0, i].set_xlabel(col)
        axes[0, i].set_ylabel('Frequency')
    
    # 2. Box plots by category
    if categorical_cols and len(numeric_cols) >= 2:
        for i, col in enumerate(numeric_cols[:2]):
            sns.boxplot(data=df, x=categorical_cols[0], y=col, ax=axes[1, i])
            axes[1, i].set_title(f'{col} by {categorical_cols[0]}')
            axes[1, i].tick_params(axis='x', rotation=45)
    
    # 3. Correlation heatmap
    corr_matrix = df[numeric_cols].corr()
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0,
                square=True, ax=axes[1, 2], cbar_kws={"shrink": .8})
    axes[1, 2].set_title('Feature Correlation Matrix')
    
    # 4. Scatter plots
    if len(numeric_cols) >= 2:
        for i in range(2):
            if i < len(numeric_cols) - 1:
                scatter_data = df if target_col is None else df
                if target_col:
                    colors = df[target_col]
                    scatter = axes[2, i].scatter(df[numeric_cols[i]], df[numeric_cols[i+1]], 
                                               c=colors, alpha=0.6, cmap='viridis')
                    plt.colorbar(scatter, ax=axes[2, i])
                else:
                    axes[2, i].scatter(df[numeric_cols[i]], df[numeric_cols[i+1]], alpha=0.6)
                
                axes[2, i].set_xlabel(numeric_cols[i])
                axes[2, i].set_ylabel(numeric_cols[i+1])
                axes[2, i].set_title(f'{numeric_cols[i]} vs {numeric_cols[i+1]}')
    
    # 5. Target distribution
    if target_col:
        df[target_col].value_counts().plot(kind='bar', ax=axes[2, 2])
        axes[2, 2].set_title(f'Distribution of {target_col}')
        axes[2, 2].set_xlabel(target_col)
        axes[2, 2].set_ylabel('Count')
    
    plt.tight_layout()
    plt.show()

# Create EDA dashboard
create_eda_dashboard(df, target_col='target')
```
