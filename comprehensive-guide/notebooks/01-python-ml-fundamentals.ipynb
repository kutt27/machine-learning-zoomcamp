## 🔧 Functions and Classes for ML

```python
# Function for data preprocessing
def preprocess_features(X, method='standardize'):
    """
    Preprocess feature matrix
    
    Parameters:
    X (np.array): Feature matrix
    method (str): 'standardize' or 'normalize'
    
    Returns:
    np.array: Preprocessed features
    """
    if method == 'standardize':
        return (X - X.mean(axis=0)) / X.std(axis=0)
    elif method == 'normalize':
        return (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))
    else:
        raise ValueError("Method must be 'standardize' or 'normalize'")

# Test the function
test_data = np.random.randn(100, 3) * 10 + 5
standardized = preprocess_features(test_data, 'standardize')
normalized = preprocess_features(test_data, 'normalize')

print(f"Original mean: {test_data.mean(axis=0)}")
print(f"Standardized mean: {standardized.mean(axis=0)}")
print(f"Normalized range: [{normalized.min():.3f}, {normalized.max():.3f}]")
```

```python
# Simple ML model class
class SimpleLinearRegression:
    def __init__(self):
        self.weights = None
        self.bias = None
        self.is_fitted = False
    
    def fit(self, X, y):
        """Fit the model using normal equation"""
        # Add bias term
        X_with_bias = np.column_stack([np.ones(X.shape[0]), X])
        
        # Normal equation: θ = (X^T X)^(-1) X^T y
        theta = np.linalg.inv(X_with_bias.T @ X_with_bias) @ X_with_bias.T @ y
        
        self.bias = theta[0]
        self.weights = theta[1:]
        self.is_fitted = True
        
        return self
    
    def predict(self, X):
        """Make predictions"""
        if not self.is_fitted:
            raise ValueError("Model must be fitted before making predictions")
        
        return X @ self.weights + self.bias
    
    def score(self, X, y):
        """Calculate R² score"""
        predictions = self.predict(X)
        ss_res = np.sum((y - predictions) ** 2)
        ss_tot = np.sum((y - np.mean(y)) ** 2)
        return 1 - (ss_res / ss_tot)

# Test the class
X_train = np.random.randn(100, 2)
y_train = X_train @ np.array([2, -1]) + np.random.randn(100) * 0.1

model = SimpleLinearRegression()
model.fit(X_train, y_train)
score = model.score(X_train, y_train)

print(f"Model weights: {model.weights}")
print(f"Model bias: {model.bias:.3f}")
print(f"R² score: {score:.3f}")
```
